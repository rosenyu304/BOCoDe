# BoCoDe algorithms — example runs (experimental mode)

This notebook runs each Bayesian-optimization baseline in `algorithms/` once, in
**experimental mode**: every run prints best-so-far each iteration (`--show_progress`)
and saves its full per-iteration trace to an `.npz` under `examples/results/`
(`--saved_full_experiment`). The demos use the synthetic test functions in
`bocode.synthetic` (Ackley, Branin, BraninCurrin) plus a couple of real problems for
the constrained demos. Run all cells top to bottom; the last cell loads a saved run.

**Conventions.** Every objective is *maximized*; reported values are the best-so-far
objective (single-objective) or the dominated hypervolume (multi-objective). Initial
designs use a SciPy Latin-Hypercube sample. Iteration counts are kept small here so
the notebook runs quickly. For batch/cluster sweeps see `run_experiments.py` and
`README.md` in this folder.
